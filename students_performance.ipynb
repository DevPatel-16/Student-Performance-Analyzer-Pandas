```python
# Handling the missing values
# For missing subject marks, consider replacing them with that subject's median
select_col = df.isnull().any(axis=1)
df[select_col]
df_clean = df.copy()

df_clean['Maths'] = df_clean['Maths'].fillna(df_clean['Maths'].median())
df_clean['Statistics'] = df_clean['Statistics'].fillna(df_clean['Statistics'].median())
df_clean['Attendance'] = df_clean['Attendance'].fillna(df_clean['Attendance'].median())
df_clean.isna().sum()  #check the missing values are handled or not
```

```python
df_final= df_clean.drop_duplicates()  #Remove duplicate records
```

```python
df_final.isna().any(axis=1)  #check for the null values
```

```python
df_final.reset_index(drop=True)
```

```python
df_final['Total'] = df[['Python','Statistics','Maths']].sum(axis=1)
df_final
```

```python
df_final['Average'] = df[['Python','Statistics','Maths']].mean(axis=1)
```

```python

conditions = [
    df_final['Average'] >= 85,
    (df_final['Average'] >= 70) & (df_final['Average'] < 85),
    (df_final['Average'] >= 50) & (df_final['Average'] < 70),
    df_final['Average'] < 50
]

grades = ['A', 'B', 'C', 'D']

df_final['Grade'] = np.select(conditions, grades, default='F')
df_final[['Name', 'Average', 'Grade']]
```

```python
condition2 = (
    (df_final['Python'] >= 40) &
    (df_final['Statistics'] >= 40) &
    (df_final['Maths'] >= 40)
)

df_final['Result'] = np.where(condition2, "Pass", "Fail")

df_final[['Name', 'Python', 'Maths', 'Statistics', 'Result']]
```
